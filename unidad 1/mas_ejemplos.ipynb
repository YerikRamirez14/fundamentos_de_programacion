```python
tabla = int(input("que tabla de multiplicar quieres"))
limiteinicial = int(input("dame el limite inicial"))
limitefinal = int(input("dame el limite final"))
while limitefinal < limiteinicial:
  print("el limite final debe ser mayor que el inicial")
  limitefinal = int(input("dame el limite final"))

incrementos = int(input("dame el numero de los incrementos"))

for i in range (limiteinicial, limitefinal, incrementos):
   print(tabla, "*", i, "=", tabla*i)
```

```python
alumnos = ["ana", "luis", "carlos", "juan"]
print(alumnos)
print(alumnos[0])
print(alumnos[1])
print(alumnos[2])
print(alumnos[5]) #Esta instruccion marcara error porque el indice o posicion no existe.
```

```python
# modificar un valor de la lista en un indice especifico
alumnos[1] = "maria luisa"
print(alumnos)

["ana","Maria Luisa", "carlos", "juan"]
```

```python
#El metodo append agrega un elemento al final de la lista
alumnos.append("Karla")
print(alumnos)
```

```python
#el metodo insert agruega un elemento a la lista en la posicion especificada
alumnos.insert(2,"cristina")
print(alumnos)
```

```python
#remove elimina un elemento especificando el valor del elemento
#debe coincidir exactamente mayusculas y minusculas
alumnos.remove("ana")
alumnos.remove("Ana") #esta marca error porque no esta bien escrito
```

```python
print(alumnos)
```

```python
#pop elimina el elemento en la posicion especificada
alumnos.pop(2)
print(alumnos)
```

```python
del alumnos[3]
print(alumnos)
```

```python
# la funcion len me permite conocer cuantos elementos hay en la lista
print("elementos de la lista de alumnos", len(alumnos))
```

```python
software = ["Python","Python y VS Code", "AutoCAD", "Solidworks", "Java"]
print(software)
```
